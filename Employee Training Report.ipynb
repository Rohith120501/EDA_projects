```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(data = data3, x = "Annual_Salary_INR", kde=True, color="skyblue", bins=20)
plt.title("Distribution of Annual Salary (INR)")
plt.xlabel("Annual Salary (INR)")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(
    data=data3, x="Department", y="Annual_Salary_INR", palette="Set2", hue="Department"
)
plt.title("Salary Distribution by Department")
plt.xlabel("Department")
plt.ylabel("Annual Salary (INR)")
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.violinplot(
    x="Education_Level",
    y="Annual_Salary_INR",
    data=data3,
    palette="Pastel1",
    hue="Education_Level",
    inner="quartile",
)
plt.title("Annual Salary Distribution by Education Level")
plt.xlabel("Education Level")
plt.ylabel("Annual Salary")
plt.xticks(rotation=45)
plt.tight_layout()
```

```python
data4 = data3.head(50)
plt.figure(figsize=(10, 6))
sns.scatterplot(
    x="Experience_Years",
    y="Training_Hours",
    hue="Attrition",
    data=data4,
    alpha=0.7,
    palette="Accent",
)
plt.title("Experience vs Training Hours")
plt.xlabel("Experience (Years)")
plt.ylabel("Training Hours")
plt.legend(title="Attrition")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.kdeplot(
    data=data3,
    x="Annual_Salary_INR",
    hue="Work_Mode",
    fill=True,
    common_norm=False,
    alpha=0.4,
    palette="muted",
)
plt.title("Salary Density Distribution by Work Mode")
plt.xlabel("Annual Salary (INR)")
plt.ylabel("Density")
plt.show()
```

```python
data5 = data3[data3["Region"] == "North"]
data6 = data3[data3["Region"] == "South"]
data7 = data3[data3["Region"] == "East"]
data8 = data3[data3["Region"] == "West"]
fig,axes = plt.subplots(2, 2, figsize=(16, 12))

sns.countplot(data=data5, x="Employment_Type", color="green", ax=axes[0, 0])
axes[0, 0].set_title("North Region - Employment Type")

sns.countplot(data=data6, x="Employment_Type", color="Red", ax=axes[0, 1])
axes[0, 1].set_title("South Region - Employment Type")

sns.countplot(data=data7, x="Employment_Type", color="Blue", ax=axes[1, 0])
axes[1, 0].set_title("East Region - Employment Type")

sns.countplot(data=data8, x="Employment_Type", color="Orange", ax=axes[1, 1])
axes[1, 1].set_title("West Region - Employment Type")
```
