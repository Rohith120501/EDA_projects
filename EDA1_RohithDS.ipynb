```python
def set_value(x):
    if x == "Books":
        return 25179.85
    elif x == "Electronics":
        return 25028.02
    elif x == "Fashion":
        return 25349.00
    elif x == "Grocery":
        return 23892.02
    else:
        return 24620.25
```

```python
df_2_nulls = df_2[df_2["Unit_Price_INR"].isnull()]
df_2_notnulls = df_2[~df_2["Unit_Price_INR"].isnull()]
```

```python
df_2_nulls["Unit_Price_INR"] = df_2_nulls["Product_Category"].apply(set_value)
```
