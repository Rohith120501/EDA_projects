```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
ds4
```

```python
yearly_mean = ds4.groupby('Enrollment_Year')['Final_Exam_Score'].mean()
yearly_mean
plt.plot(yearly_mean.index,yearly_mean.values)
plt.xlabel("Year")
plt.ylabel("Average Score")
```

```python
plt.figure(figsize=(6,4))
plt.title("Avg Score by Placements")
sns.lineplot(data = ds4,
             x = "Placement_Status",
             y = "Final_Exam_Score",
             color= 'red',
             marker = "o",
             hue = "Club_Participation")
```

```python
plt.figure(figsize=(10,8))
plt.title ("Attendance to Final Score")
sns.scatterplot(data = ds4,
             x = "Study_Hours_Per_Day",
             y = "Final_Exam_Score",hue = "Placement_Status",
             color= 'red')
plt.xlabel("Study Hours per day")
plt.ylabel("Final Score")
```

```python
plt.figure(figsize=(8, 6))
numeric_cols = [
    "Age",
    "Year_of_Study",
    "Semester",
    "Attendance_Percent",
    "Study_Hours_Per_Day",
    "Assignment_Score",
    "Midterm_Score",
    "Final_Exam_Score",
]
corr = ds4[numeric_cols].corr()
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)
plt.title("Correlation Heatmap of Student Numerical Features")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(20, 8))
sns.countplot( data = ds4,x = "Department",hue = 'Placement_Status')
plt.title("Student Count by Department and Placement Status")
plt.xlabel("Department")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(
    data=ds4,
    x="Final_Exam_Score",
    hue="Club_Participation",
    kde=True,
    multiple="stack",
    palette="viridis",
)
plt.title("Final Exam Scores by Club Participation")
plt.xlabel("Final Exam Score")
plt.ylabel("Count")
plt.show()
```
