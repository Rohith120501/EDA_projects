```python
for i in data.columns:
    print(f"{i}")
    q1 = data[i].quantile(0.25)
    q3 = data[i].quantile(0.75)
    IQR = q3-q1
    print(f"IQR: {IQR}")
    LB = q1 - 1.5 * IQR
    UB = q3 + 1.5 * IQR
    print(f"Lower Bound: {LB}\nUpper Bound: {UB}")
    outlier = data[(data[i] < LB) | (data[i] > UB)]
    print("Outlier is:", outlier)
    print("\n")
    print("\n")
    
```
